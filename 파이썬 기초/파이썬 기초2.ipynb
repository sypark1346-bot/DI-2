### 03-2 while문

```python
treeHit = 1
while treeHit <= 10:
    print(f"나무를 {treeHit}번 찍었습니다.")
    treeHit = treeHit + 1
```

### 03-03 for문

```python
pocket = ['iphone', 'money', 'card', 'key', 'trash']
for item in pocket:
    print(item)
```

```python
scores = [90, 50, 70, 65, 75]

for score in scores:
    if score >= 60:
        print(f"{score}점 합격입니다.")
    else:
        print(f"{score}점 불합격입니다.")
```

### 리스트 컴프리헨션

```python
result = [num * 5 for num in range(1, 11)]
print(result)   # [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
```

```python
a = []                     # 1. 빈 리스트를 먼저 만들기
for i in range(1, 11):
    a.append(i * 5)        # 2. 계산해서 하나씩 넣기
print(a)                   # 3. 반복이 끝난 뒤 출력
```

```python
my_bag = ['book', 'pouch', 'notebook', 'chocolate', 'chips']
for i, item in enumerate(my_bag):
    print(f"{i+1}번째 꺼낸 것은 {item}입니다.")
```
