## ✅ 문제 4. 추천 항목 추가 함수 만들기

- `add_recommendation(mbti, category, item)` 함수를 만듭니다.
    - 함수명과 변수명을 적당하게 바꿔도 좋습니다.
- 해당 MBTI가 없으면 새로 만들고, 카테고리가 없다면 새로 추가합니다.

````python
def add_recommendation(mbti, category, item):
    # 1. MBTI가 없으면 새 키로 추가
    if mbti not in recommendations:
        recommendations[mbti] = []

    # 2. 카테고리를 찾아서, 있으면 항목 덧붙이기
    for cat, items in recommendations[mbti]:
        if cat == category:
            items.append(item)
            return

    # 3. 카테고리가 없으면 (카테고리, [항목]) 튜플 추가
    recommendations[mbti].append((category, [item]))

personality = {
    'INTJ': ['전략적인', '독립적인'],
    'INTP': ['논리적인', '호기심 많은'],
    'ENTJ': ['리더십 있는', '결단력 있는'],
    'ENTP': ['재치 있는', '도전적인'],
    'INFJ': ['통찰력 있는', '배려심 깊은'],
    'INFP': ['감성적인', '이상주의적인'],
    'ENFJ': ['다정한', '사교적인'],
    'ENFP': ['열정적인', '창의적인'],
    'ISTJ': ['책임감 있는', '꼼꼼한'],
    'ISFJ': ['헌신적인', '따뜻한'],
    'ESTJ': ['체계적인', '현실적인'],
    'ESFJ': ['친절한', '협조적인'],
    'ISTP': ['침착한', '실용적인'],
    'ISFP': ['온화한', '예술적인'],
    'ESTP': ['활동적인', '대담한'],
    'ESFP': ['유쾌한', '자유로운'],
}

compatibility = {
    'INTJ': ['ENFP', 'ENTP'],
    'INTP': ['ENTJ', 'ESTJ'],
    'ENTJ': ['INFP', 'INTP'],
    'ENTP': ['INFJ', 'INTJ'],
    'INFJ': ['ENFP', 'ENTP'],
    'INFP': ['ENFJ', 'ENTJ'],
    'ENFJ': ['INFP', 'ISFP'],
    'ENFP': ['INFJ', 'INTJ'],
    'ISTJ': ['ESFP', 'ESTP'],
    'ISFJ': ['ESFP', 'ESTP'],
    'ESTJ': ['ISFP', 'ISTP'],
    'ESFJ': ['ISFP', 'ISTP'],
    'ISTP': ['ESFJ', 'ESTJ'],
    'ISFP': ['ENFJ', 'ESFJ'],
    'ESTP': ['ISFJ', 'ISTJ'],
    'ESFP': ['ISFJ', 'ISTJ'],
}

for mbti, partners in compatibility.items():
    for partner in partners:
        add_recommendation(mbti, '궁합', partner)

for mbti, traits in personality.items():
    for trait in traits:
        add_recommendation(mbti, '성격', trait)

print(recommendations)
````

- `recommendations` 딕셔너리에 MBTI가 **없는 경우**, 새로운 키로 추가해야 합니다.
- MBTI가 있다면, 그 안의 리스트에서 **해당 카테고리를 찾아야** 합니다.
- 이미 존재하는 카테고리라면, 항목을 그 리스트에 **덧붙이는 방식(append)**으로 추가합니다.
- 카테고리가 없다면, `(카테고리, [항목])` 형태의 튜플을 리스트에 추가해보는 건 어떨까요?

- 반복문을 통해 모든 사용자의 추천을 출력합니다.

````python
def print_recommendation(mbti):
    if mbti not in recommendations:
        print(f'{mbti}에 대한 추천이 없어요.')
        return

    for category, items in recommendations[mbti]:
        print(f'{category}: {", ".join(items)}')


users = []

# 1. 사용자 추가
users.append(('민지', 'INFP'))
users.append(('윤희', 'ESTJ'))
users.append(('정훈', 'ENFP'))

# 2. 모든 사용자의 추천 출력
for user in users:
    name, mbti = user
    print(f'\n[{name}님 ({mbti})의 추천]')
    print_recommendation(mbti)
````

## ✅ 문제 6. 콘솔 앱처럼 메뉴 구성하기

- 아래와 같은 메뉴를 출력하고, 사용자로부터 반복적으로 입력을 받아 기능을 수행하는 프로그램을 만들어봅시다.
- `while True` 반복문과 `if` 조건문을 활용합니다.
- 각 기능은 함수로 따로 만들어 호출하는 것이 좋습니다. 위에서 만든 함수를 활용해볼까요?
- 사용자가 4번을 선택하면 프로그램이 종료되어야 합니다.

> 출력 결과 예시:
```
[ MBTI 취향 찾기 메뉴 ]
1. 추천 보기
2. 추천 추가
3. 사용자 목록 보기
4. 종료
```
```
번호를 입력하세요: 1
```

````python
def print_recommendation(mbti):
    if mbti not in recommendations:
        print(f'{mbti}에 대한 추천이 없어요.')
        return

    for category, items in recommendations[mbti]:
        print(f'{category}: {", ".join(items)}')


def show_menu():
    print('\n[ MBTI 취향 찾기 메뉴 ]')
    print('1. 추천 보기')
    print('2. 추천 추가')
    print('3. 사용자 목록 보기')
    print('4. 종료')


def show_users():
    if not users:
        print('등록된 사용자가 없어요.')
        return

    for name, mbti in users:
        print(f'{name} ({mbti})')


def main():
    while True:
        show_menu()
        choice = input('번호를 입력하세요: ')
        print(f'번호를 입력하세요: {choice}')

        if choice == '1':
            mbti = input('MBTI를 입력하세요: ').strip().upper()
            print(f'MBTI를 입력하세요: {mbti}')
            print_recommendation(mbti)

        elif choice == '2':
            mbti = input('MBTI를 입력하세요: ').strip().upper()
            print(f'MBTI를 입력하세요: {mbti}')
            category = input('카테고리를 입력하세요 (예: 취미, 음악): ')
            print(f'카테고리: {category}')
            item = input('추가할 항목을 입력하세요: ')
            print(f'추가할 항목: {item}')
            add_recommendation(mbti, category, item)
            print('추가했어요!')
            print_recommendation(mbti)

        elif choice == '3':
            show_users()

        elif choice == '4':
            print('프로그램을 종료합니다.')
            break

        else:
            print('1~4 중에서 선택해 주세요.')


# 프로그램 실행
main()
````

- 메뉴 번호가 1이면 추천 보기, 2면 추천 추가, 3이면 사용자 목록 보기, 4면 종료입니다.
