## Q1. 월별 매출 추이는?

주문일을 월 단위로 묶어(`date_trunc`) 매출 주문(Paid·Shipped)의 건수와 매출 합계를 구합니다.

```python
q1 = run("""
SELECT
  date_trunc('month', order_date) AS month,
  COUNT(*)                        AS orders,
  SUM(amount)                     AS revenue,
  ROUND(AVG(amount))              AS avg_order_value   -- 객단가 (주문 1건당 평균 금액)
FROM project_name.dataset_name.orders
WHERE
  status IN ('Paid', 'Shipped')     -- 실제 매출 주문만
GROUP BY
  month
ORDER BY
  month                             -- 시간 순서대로 봐야 추이가 보임
""")
q1
```

```python
plt.figure(figsize=(9, 4))
plt.plot(q1['month'], q1['revenue'], marker='o')
plt.title('Monthly Revenue (Paid + Shipped)')
plt.xlabel('Month')
plt.ylabel('Revenue (KRW)')
plt.grid(True, alpha=0.3)
plt.show()
```

매출이 가장 높았던 달 TOP 3를 뽑아 봅니다. 매출 큰 순으로 정렬(`DESC`)한 뒤 위에서 3개만 자릅니다(`LIMIT`).

```python
run("""
SELECT
  date_trunc('month', order_date) AS month,
  SUM(amount)                     AS revenue
FROM project_name.dataset_name.orders
WHERE
  status IN ('Paid', 'Shipped')
GROUP BY
  month
ORDER BY
  revenue DESC                      -- 매출 큰 달부터
LIMIT 3                             -- 상위 3개만
""")
```

**💡 인사이트**: 월 주문 건수는 매달 4건으로 같은데 매출은 48.8만~85만 원으로 크게 출렁입니다. 매출 차이는 **주문 수가 아니라 객단가(고액 주문 여부)** 에서 나옵니다. (2024년 2월은 11일까지만 데이터가 있어 비교에서 제외)

## Q2. 주문 중 정상 / 결제대기 / 이탈(취소·반품) 비율은?

먼저 `DISTINCT`로 주문 상태에 어떤 값들이 있는지 확인합니다.

```python
run("""
SELECT DISTINCT
  status
FROM project_name.dataset_name.orders
ORDER BY
  status
""")
```

상태 값은 5가지입니다. `CASE WHEN`으로 이를 3개 그룹으로 묶고, 그룹별 건수와 전체 대비 비율을 구합니다.  
비율의 분모(전체 주문 수)는 서브쿼리로 구합니다.

```python
q2 = run("""
SELECT
  CASE
    WHEN status IN ('Paid', 'Shipped')       THEN '정상'
    WHEN status = 'Placed'                   THEN '결제대기'
    WHEN status IN ('Cancelled', 'Returned') THEN '이탈(취소·반품)'
  END                                          AS order_group,
  COUNT(*)                                     AS orders,
  ROUND(COUNT(*) * 100.0
        / (SELECT COUNT(*) FROM project_name.dataset_name.orders), 1) AS pct
FROM project_name.dataset_name.orders
GROUP BY
  order_group
ORDER BY
  orders DESC                       -- 많은 그룹부터
""")
q2
```

**💡 인사이트**: 전체 30건 중 6건(20%)이 취소·반품으로 이탈했습니다. **주문 5건 중 1건이 매출로 이어지지 못하므로**, 취소·반품 사유를 확인해 줄이는 것이 매출을 늘리는 쉬운 방법입니다.

## Q3. 2번 이상 구매한 재구매 고객은 누구이고, 얼마나 샀나?

고객별로 매출 주문을 묶은 뒤, **`HAVING`** 으로 "주문 2건 이상"인 그룹만 남깁니다.  
(`WHERE`는 묶기 전의 행을 거르고, `HAVING`은 묶은 뒤의 그룹을 거릅니다.)

```python
q3 = run("""
SELECT
  customer_id,
  COUNT(*)    AS orders,
  SUM(amount) AS total_amount
FROM project_name.dataset_name.orders
WHERE
  status IN ('Paid', 'Shipped')
GROUP BY
  customer_id
HAVING
  COUNT(*) >= 2                     -- 재구매 고객만
ORDER BY
  total_amount DESC                 -- 많이 산 고객부터
""")
q3
```

```python
# 비교용: 매출 주문이 있는 전체 고객 수와 전체 매출
total = run("""
SELECT
  COUNT(DISTINCT customer_id) AS buyers,
  SUM(amount)                 AS total_revenue
FROM project_name.dataset_name.orders
WHERE
  status IN ('Paid', 'Shipped')
""")
print('재구매 고객 수:', len(q3), '/ 구매 고객 수:', total['buyers'][0])
print('재구매 고객 매출 비중:', round(q3['total_amount'].sum() / total['total_revenue'][0] * 100, 1), '%')
```

**💡 인사이트**: 구매 고객 12명 중 8명이 재구매 고객이고, 이들이 **전체 매출의 약 89%** 를 만듭니다. 신규 유치보다 기존 고객을 붙잡는 것(재구매 혜택 등)이 매출에 더 중요합니다.

## Q4. 카테고리별 가격대는?

상품을 카테고리별로 묶어 평균·최저·최고 가격을 구하고, 평균 가격이 높은 순으로 정렬합니다.

```python
q4 = run("""
SELECT
  category,
  COUNT(*)            AS products,
  ROUND(AVG(price))   AS avg_price,
  MIN(price)          AS min_price,
  MAX(price)          AS max_price
FROM project_name.dataset_name.products
GROUP BY
  category
ORDER BY
  avg_price DESC                    -- 비싼 카테고리부터
""")
q4
```

가장 비싼 카테고리인 부츠 상품을 상품명으로 찾아 봅니다. `LIKE '%부츠%'`는 이름 어딘가에 '부츠'가 들어간 상품을 찾습니다.

```python
run("""
SELECT
  product_name,
  category,
  price
FROM project_name.dataset_name.products
WHERE
  product_name LIKE '%부츠%'        -- 이름에 '부츠'가 들어간 상품
ORDER BY
  price DESC
""")
```

**💡 인사이트**: Boots 평균가(16만 원)가 Accessory(8,500원)의 약 19배입니다. Boots·Running은 **객단가를 끌어올리는 주력 상품**, Accessory는 다른 상품과 함께 사는 **추가 구매용**으로 활용할 수 있습니다. (주문 데이터에 상품 정보가 없어 카테고리별 실제 판매량은 확인할 수 없습니다.)

## Q5. 국가별 고객 수와 등급 구성은?

국가별로 고객을 묶고, `CASE WHEN`으로 등급별 인원을 각각 셉니다.  
조건에 맞으면 1, 아니면 NULL이 되고 `COUNT`는 NULL을 세지 않으므로 조건에 맞는 사람 수만 나옵니다.  
국가가 비어 있는(NULL) 고객은 `COALESCE`로 '미입력'으로 표시합니다.

```python
q5 = run("""
SELECT
  COALESCE(country, '미입력')                         AS country,
  COUNT(*)                                            AS customers,
  COUNT(CASE WHEN grade = 'Gold'   THEN 1 END)        AS gold,
  COUNT(CASE WHEN grade = 'Silver' THEN 1 END)        AS silver,
  COUNT(CASE WHEN grade = 'Bronze' THEN 1 END)        AS bronze,
  COUNT(CASE WHEN grade IS NULL    THEN 1 END)        AS no_grade
FROM project_name.dataset_name.customers
GROUP BY
  country
ORDER BY
  customers DESC,                   -- 고객 많은 나라부터
  gold DESC                         -- 같으면 Gold 많은 나라 먼저
""")
q5
```

**💡 인사이트**: 고객 15명 중 8명(53%)이 한국 고객입니다. 미국은 고객 3명 중 **등급이 있는 2명이 모두 Gold**라 인원 대비 우수 고객 비중이 가장 높고, 일본은 Gold가 없습니다. 국가·등급이 비어 있는 고객도 1명씩 있어 데이터 입력 보완이 필요합니다.
